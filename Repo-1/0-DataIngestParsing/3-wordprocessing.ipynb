### Word Document Processing ###

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
print("Loading DOCX file using Docx2txtLoader...")
# Load the DOCX file using Docx2txtLoader   
# https://reference.langchain.com/python/langchain-community/document_loaders/word_document/Docx2txtLoader
try:
    loader1 = Docx2txtLoader("data/docx/conversation_Schema.docx")
    documents1 = loader1.load()
    print("Documents loaded using Docx2txtLoader:")
    print(documents1)
    print(f"Number of documents loaded: {len(documents1)}")
    print("--------------------------------------------------")
    print(f"Content of the first document: {documents1[0].page_content[:200]}")  # Print the first 200 characters of the first document
    print("--------------------------------------------------")
    print(f"Metadata of the first document: {documents1[0].metadata}")
except Exception as e:
    print(f"Error occurred while loading DOCX file: {e}")
```

```python
##Method2

print("Loading DOCX file using unstructuredDocxLoader...")
# Load the DOCX file using unstructuredDocxLoader
# https://reference.langchain.com/python/langchain-community/document_loaders/word_document/UnstructuredWordDocumentLoader
try:
    loader2= UnstructuredWordDocumentLoader("data/docx/conversation_Schema.docx",mode="elements")
    documents2 = loader2.load()
    print("Documents loaded using unstructuredWordDocumentLoader:")
    print(documents2)
    print(f"Number of documents loaded: {len(documents2)}")
    print("--------------------------------------------------")
    for i,doc in enumerate(documents2[:3]):
        print(f"Content of document {i+1}: {doc.page_content[:200]}")  # Print the first 200 characters of the first three documents
        print("--------------------------------------------------")
        print(f"Metadata of document {i+1}: {doc.metadata.get('category', 'No metadata available')}")
        print("--------------------------------------------------")
except Exception as e:
    print(f"Error occurred while loading DOCX file: {e}")
```
